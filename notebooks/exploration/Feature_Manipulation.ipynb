```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

import seaborn as sns
```

```python
filename = r"../../data/raw/term-deposit-marketing-2020.csv"
df = pd.read_csv(filename)
```

# making a no outlier month dataset

```python
no_outlier_df = df.copy().replace("unknown", pd.NA).dropna()

no_outlier_df = no_outlier_df.replace("may", pd.NA).dropna()
no_outlier_df = no_outlier_df.replace("oct", pd.NA).dropna()
no_outlier_df = no_outlier_df.replace("dec", pd.NA).dropna()
no_outlier_df.to_csv(r"../../data/processed/no_month_outliers_no_unknown.csv")
```

# Age - binning by generation

```python
new_df = df.copy()
df.columns
```

```python
# making generation boundaries
import datetime
current_year = datetime.date.today().year

# one older than greatest generation is lumped in with greatest generation
greatest_generation_youngest = current_year - 1927
silent_generation_youngest = current_year - 1945
baby_boomers_generation_youngest = current_year - 1964
gen_x_generation_youngest = current_year - 1980
millennial_generation_youngest = current_year - 1996
gen_z_generation_youngest = current_year - 2012
alpha_generation_youngest = current_year - 2024

boundaries = [0, 
            alpha_generation_youngest, 
            gen_z_generation_youngest, 
            millennial_generation_youngest, 
            gen_x_generation_youngest, 
            baby_boomers_generation_youngest, 
            silent_generation_youngest, 
            greatest_generation_youngest,
            150]

group_names = [
            "gen_beta",
            "gen_alpha",
            "gen_z",
            "millennial",
            "gen_x",
            "baby_boomers",
            "silent_generation",
            "greatest_generation",
               ]

new_df["age"] = pd.cut(new_df["age"], bins=boundaries, labels = group_names, include_lowest=True)
new_df = new_df.replace("unknown", pd.NA).dropna()
```

```python
new_df.to_csv(r"../../data/processed/binned_age.csv")
```
