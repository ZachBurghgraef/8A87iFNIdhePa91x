# Feature Distributions Based on the y

```python
num_cols = df.select_dtypes(include=['number']).columns
cat_cols = df.select_dtypes(include=['object', 'str', 'category']).columns.drop('y')

all_cols = list(num_cols) + list(cat_cols)
n_plots = len(all_cols)

for col in all_cols:
    plt.figure(figsize=(10,5))
    if col in num_cols:
        sns.histplot(data=df, x=col, hue='y', stat='density', kde=True)
    else:
        sns.countplot(data=df, x=col, hue='y')
    plt.title(f'{col} by y')
    plt.tight_layout()
    plt.show()
```
